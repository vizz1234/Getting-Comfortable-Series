## 1. Gemini Chat Models

Let's initialize the Gemini Chat Model and make a simple call.

```python
from langchain_google_genai import ChatGoogleGenerativeAI

# Initialize Gemini 1.5 Flash
model = ChatGoogleGenerativeAI(model="gemini-1.5-flash", temperature=0.7)

response = model.invoke("Explain LLM orchestration in one short sentence.")
print(response.content)
```

## 2. Prompt Templates

Prompt templates help organize instructions and user input.

```python
from langchain_core.prompts import ChatPromptTemplate

prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant that explains {topic} in a {tone} tone."),
    ("user", "{query}")
])

# Format the prompt to see what it looks like
prompt_value = prompt_template.invoke({
    "topic": "quantum computing",
    "tone": "pirate",
    "query": "What is a qubit?"
})
prompt_value
```

## 3. Chaining with LCEL (LangChain Expression Language)

Using the pipe `|` operator, we can chain our prompt template directly to our model and optionally output parsers.

```python
from langchain_core.output_parsers import StrOutputParser

chain = prompt_template | model | StrOutputParser()

result = chain.invoke({
    "topic": "databases",
    "tone": "medieval knight",
    "query": "Explain indexing to me."
})
print(result)
```
